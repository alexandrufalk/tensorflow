## What is generative AI?
Generative AI(Gen AI) is a subset of artificial intelligence(AI) that uses generative models to produce new text, images, videos and other forms of data. These networks are adept at identifying patterns within the training data, which allows them to produce new, similar data points that maintain the statistical properties of the original dataset. Gen AI is widely used in industry such as for creating realistic CGI and virtual environments in entertainment and movie productions. The healthcare sector leverages generative AI for drug discovery and enhancing medical imaging. In finance, it’s used for simulating market scenarios and generating synthetic data for stress testing etc

## What is a Large Language Model(LLM)?
LLMs are a category of foundation models characterized by their extensive parameter count, often reaching into the billions. These models undergo pretraining on vast datasets, which equips them with the ability to comprehend and produce natural language, as well as other forms of content. LLMs excel in tasks involving language understanding and generation, making them a powerful tool in various applications. Their capacity to learn from massive amounts of data allows them to generate human-like text and perform complex language-related tasks with high proficiency.

### Architecture of LLMs: Introductory concepts

**Transformer**: <br> Transformer is a deep learning model and the foundational block of LLMs such as BERT and GPT. It is a sequence-to-sequence architecture based model that uses _**attention**_ mechanism to capture context information in the input sequence. Some significant components in the transformer are:
* **Tokenization and Input embeddings**: Tokenization is the process of converting a sequence of text into smaller unit suchs as words or subwords. The input sequence to the transformer is tokenized and mapped to a high-dimensional vector space to obtain input embeddings.
* **Encoder**: The encoder processes the input embeddings, incorporating positional encodings to maintain the order of tokens. It computes attention weights to determine the significance of each token relative to others, generating a hidden state that encapsulates contextual information.
* **Decoder**: This hidden state information is passed to decoder block to generate the target output. The decoder predicts probabilities over different words to predict a output sequence that is coherent to the learned context.

## Example 1: Apply pretrained LLMs on NLP tasks using Hugging Face Transformers : Pipelines

Hugging Face Transformers **Pipelines** are objects that abstract most of the complex code from the Transformers library, offering a simple API dedicated to tasks. The list of currently accepted tasks are : https://huggingface.co/docs/transformers/v4.46.2/en/main_classes/pipelines#transformers.pipeline.task

<center><img src="work/session_1/hfnlp_pipeline.jpg" width="700" height="300"></center>

```python
# Import libraries
import os
import pandas as pd
from transformers import pipeline

# Setup parameters for jupyter notebook
import warnings
pd.set_option('max_colwidth', None)
warnings.filterwarnings('ignore')

# Setup certificate permissions to access external API from Continental's VPN
os.environ['REQUESTS_CA_BUNDLE']  = "work/certificates.pem"
```

```python
# Example: Text classification for sentiment analysis
# Apply LLM to determine the emotional tone, sentiment or attitude expressed in a text.
# Sentiments used for this example are positive and negative

sentiment_pipeline = pipeline(
    "sentiment-analysis",
    model="distilbert/distilbert-base-uncased-finetuned-sst-2-english",
    trust_remote_code=True
)

data = [
    "Sunrise is a beautiful time of the day",
    "Cloudy days can be gloomy and reduce productivity."
]

print(sentiment_pipeline(data))
```

```python
# Example: Knowledge distillation for question answering
# Apply LLM to comprehend a given context and generate precise answers to questions related to that context.
# The process involves understanding the question, identifying relevant information within the context, and
# formulating an accurate response.

oracle = pipeline(
    model="distilbert/distilbert-base-cased-distilled-squad"
)

print(oracle(question="Where do I live?", context="My name is Wolfgang and I live in Berlin"))
```

```python
# Example: Knowledge distillation for summarization of text
# Apply LLM to generate concise summaries of given text samples. This involves analyzing the input text,
# identifying key points and main ideas, and producing a coherent summary that captures the essence of
# the original content.

text_to_summarize = "Advantages of AI: AI drives down the time taken to perform a task. It enables multi-tasking \
and eases the workload for existing resources. AI enables the execution of hitherto complex tasks without significant \
cost outlays. AI operates 24x7 without interruption or breaks and has no downtime. AI augments the capabilities of \
differently abled individuals. AI has mass market potential, it can be deployed across industries."

summarizer = pipeline(
    "summarization",
    model="google/flan-t5-small"
)

print(summarizer(text_to_summarize, min_length=5, max_length=20))
```

```python
# Example: Zero shot classification
# Apply LLM to for zero-shot text classification, categorizing sample text into predefined categories
# without the need for training examples. LLMs leverage their pre-trained knowledge to understand the
# text and assign it to the most appropriate category. This method is suitable when labeled data is scarce or unavailable

classifier = pipeline(
    "zero-shot-classification",
    model="distilbert/distilbert-base-uncased-finetuned-sst-2-english"
)

classifier(
    "This is a course about the Transformers library",
    candidate_labels=["education", "politics", "business"],
)
```

```python
# Pipeline batching
# Apply a task pipeline on a dataset

from datasets import load_dataset
from transformers.pipelines.pt_utils import KeyDataset
from tqdm.auto import tqdm

dataset = load_dataset("stanfordnlp/imdb")

summarizer = pipeline(
    "summarization",
    model="google/flan-t5-small",
    truncation=True
)

# Apply declared pipeline on a small slice of loaded dataset for demonstration purposes
for out in tqdm(summarizer(KeyDataset(dataset["unsupervised"], "text")[:10], max_length=50)):
    print(out)
```

### **Finetuning GPT-2 model with Hugging Face**

GPT-2 is an open source "decoder-only" **causal language model** designed to generate new pieces of text.

```python
import time
from datasets import load_dataset
import torch
import transformers
from transformers import AutoTokenizer, AutoModelForCausalLM
from transformers import TrainingArguments, Trainer
from trl import SFTTrainer
```

```python
# Define base model that will be finetuned to custom dataset

model_id = "openai-community/gpt2"

tokenizer = AutoTokenizer.from_pretrained(model_id)
pretrained_model = AutoModelForCausalLM.from_pretrained(model_id)
```

```python
# The output from the pretrained model prior to fine-tuning for the task: text completion
# LLM is provided with an initial text prompt, and the model then attempts to complete the text in a coherent
# and contextually appropriate way

text = "Quote: Imagination is more"
inputs = tokenizer(text, return_tensors="pt")

outputs = pretrained_model.generate(**inputs, max_new_tokens=40)
print(tokenizer.decode(outputs[0], skip_special_tokens=True))
```

```python
# Load dataset from the Hugging Face Hub for finetuning the model
dataset = load_dataset("Abirate/english_quotes")

# View sample rows from dataset
pd.DataFrame.from_dict(dataset["train"]).iloc[:5,:2]
```

```python
# Define how we'd like our outputs to look like after fine-tuning.

def formatting_func(example):
    text = f"Quote: {example['quote'][0]}\nAuthor: {example['author'][0]}<eos>"
    return [text]
```

```python
# Fine-tune the model

MAX_STEPS = 10

training_args = transformers.TrainingArguments(
    per_device_train_batch_size=1,
    gradient_accumulation_steps=4,
    warmup_steps=2,
    max_steps=MAX_STEPS,
    learning_rate=2e-4,
    logging_steps=1,
    output_dir="outputs",
)

trainer = SFTTrainer(
    model=pretrained_model,
    train_dataset=dataset["train"],
    args=training_args,
    formatting_func=formatting_func,
)

trainer.train()
```

```python
# Ask the model for text completion after fine-tuning.
# Mind that now the model nicely adheres to the desired output format.

text = "Quote: Imagination is more"
inputs = tokenizer(text, return_tensors="pt")

finetuned_model = AutoModelForCausalLM.from_pretrained(f"outputs/checkpoint-{MAX_STEPS}")
outputs = finetuned_model.generate(**inputs, max_new_tokens=40)
print(tokenizer.decode(outputs[0], skip_special_tokens=True))
```
